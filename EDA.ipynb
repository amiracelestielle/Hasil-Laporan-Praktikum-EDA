Latihan 1: Operasi Dasar NumPy Array

```python
import numpy as np

harga = np.array ([5000, 7000, 3000, 12000, 4500])
print('Rata-rata harga:', harga.mean())
print('Harga tertinggi:', harga.max())
print('Harga setelah diskon 10%:', harga * 0.9)

#Tugas Analisis 1: Kalau [5000,7000,3000] * 0.9 dicoba pake list biasa, langsung error. Soalnya kalau list dikaliin angka, Python nganggepnya "gandain list-nya", bukan ngaliin tiap isinya, dan itu pun cuma bisa kalau dikali angka bulat, bukan desimal. Makanya beda sama NumPy array yang emang didesain buat ngaliin tiap elemen sekaligus tanpa perlu di-loop.
```

Latihan 2: Membuat Series & DataFrame

```python
import pandas as pd

data_kantin = {
'menu': ['Nasi Goreng', 'Es Teh', 'Mie Ayam', 'Es Teh', None],
'harga': [12000, 4000, 10000, 4000, 8000],
'terjual': [23, 40, None, 35, 18]
}

df = pd.DataFrame(data_kantin)
print(df)

#Tugas Analisis 2: Yang keliatan kosong itu kolom menu (baris terakhir) sama terjual (baris ke-3). Risikonya kalau langsung dianalisis tanpa dibersihin, hasil hitungannya bisa salah atau malah error, soalnya data yang kosong gak bisa dihitung.
```

Latihan 3: Data Loading & Inspection

```python
df = pd.read_csv('data_kantin.csv')

print(df.head()) # 5 baris pertama
print(df.info()) # tipe data & jumlah non-nul l tiap kolom
print(df.describe()) # statistik ringkas kolom numerik
print(df.shape) # jumlah (baris, kolom)

#Tugas Analisis 3: Dari df.info(), kolom menu sama terjual jumlah non-null-nya lebih dikit dari total baris. Artinya di dua kolom itu ada data yang bolong/hilang, jadi dataset ini emang belum bersih dan perlu dicek ulang sebelum diolah lebih lanjut.
```

Latihan 4: Menangani Missing Value

```python
print(df.isnull().sum()) # jumlah data kosong tiap kolom

df['terjual'] = df['terjual'].fillna(0) # isi kekosongan dengan 0
df = df.dropna(subset=['menu']) # hapus baris jika kolom menu kosong

#Tugas Analisis 4: Kolom terjual diisi pake fillna(0) soalnya kalau gak ada catatan jumlah terjualnya, wajar dianggep aja belum ada penjualan (0). Beda sama baris yang menu-nya kosong, itu langsung dihapus (dropna), soalnya kalau nama menunya aja gak ketauan, datanya jadi gak jelas dan gak ada gunanya buat dianalisis.
```

Latihan 5: Menangani Duplikat dan Tipe Data

```python
print(df.duplicated().sum()) # jumlah baris duplikat
df = df.drop_duplicates()

df['harga'] = df['harga'].astype(int) # memastikan tipe data harga adalah integer
print(df.dtypes)

#Tugas Analisis 5: Abis drop_duplicates(), jumlah barisnya berkurang karena baris yang sama persis dibuang, sisa yang unik doang. Penting banget mastiin tipe data bener dulu sebelum dianalisis, soalnya kalau misal harga masih kebaca teks (bukan angka), kita gak bisa langsung ngitung-ngitung, bisa error atau malah hasilnya ngaco.
```

Latihan 6: Data Manipulation (Filtering, Sorting, Groupby)

```python
laris = df[df['terjual'] > 20] # filtering
urut = df.sort_values(by='terjual', ascending=False) # sorting

df['total_pendapatan'] = df['harga'] * df['terjual'] # kolom turunan

ringkasan = df.groupby('menu')['total_pendapatan'].sum() # agregasi
print(ringkasan)

#Tugas Analisis 6: Dari hasil groupby, menu yang pendapatannya paling gede itu Nasi Goreng. Info ini bisa dipake kantin buat mutusin, misal nambah stok bahan buat menu yang paling laku, atau ngecek ulang menu yang pendapatannya kecil, apa emang kurang laku atau datanya yang bermasalah.
```
